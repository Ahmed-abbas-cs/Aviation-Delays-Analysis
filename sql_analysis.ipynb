```python
import pandas as pd
import sqlite3

# 1. Load the cleaned dataset
df = pd.read_csv('flights_final_project.csv')

# 2. Create a local SQLite database and load the data into a 'flights' table
conn = sqlite3.connect('flights_database.db')
df.to_sql('flights', conn, if_exists='replace', index=False)

# --- Query 1: Average arrival delay per airline ---
query_1 = """
SELECT AIRLINE, 
       COUNT(*) as Total_Flights,
       ROUND(AVG(ARR_DELAY), 2) as Avg_Arrival_Delay
FROM flights
GROUP BY AIRLINE
ORDER BY Avg_Arrival_Delay DESC
LIMIT 5;
"""
df_airlines_delay = pd.read_sql_query(query_1, conn)

# --- Query 2: Total delay minutes by root cause ---
query_2 = """
SELECT ROUND(SUM(DELAY_DUE_CARRIER), 0) as Carrier_Delay,
       ROUND(SUM(DELAY_DUE_WEATHER), 0) as Weather_Delay,
       ROUND(SUM(DELAY_DUE_NAS), 0) as Air_Traffic_Delay,
       ROUND(SUM(DELAY_DUE_SECURITY), 0) as Security_Delay,
       ROUND(SUM(DELAY_DUE_LATE_AIRCRAFT), 0) as Late_Aircraft_Delay
FROM flights;
"""
df_delay_reasons = pd.read_sql_query(query_2, conn)

# --- Query 3: Top 10 busiest airports and average departure delay ---
query_3 = """
SELECT ORIGIN as Airport, 
       COUNT(*) as Total_Flights,
       ROUND(AVG(DEP_DELAY), 2) as Avg_Departure_Delay
FROM flights
GROUP BY ORIGIN
ORDER BY Total_Flights DESC
LIMIT 10;
"""
df_busiest_airports = pd.read_sql_query(query_3, conn)

# Close the database connection
conn.close()
```
