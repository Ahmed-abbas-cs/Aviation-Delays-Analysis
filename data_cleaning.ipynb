```python
import pandas as pd

# 1. Load the raw dataset
df = pd.read_csv('flights_data.csv')

# 2. Sample 50,000 random flights for optimal performance
df_sample = df.sample(n=50000, random_state=42)

# 3. Define the essential columns for analysis
columns_to_keep = [
    'FL_DATE', 'AIRLINE', 'ORIGIN', 'DEST', 
    'DEP_DELAY', 'ARR_DELAY', 'CANCELLED',
    'DELAY_DUE_CARRIER', 'DELAY_DUE_WEATHER', 
    'DELAY_DUE_NAS', 'DELAY_DUE_SECURITY', 'DELAY_DUE_LATE_AIRCRAFT'
]

# Filter the dataset
df_clean = df_sample[columns_to_keep].copy()

# 4. Handle missing values (Nulls) in delay columns by filling them with 0
delay_columns = [
    'DEP_DELAY', 'ARR_DELAY', 'DELAY_DUE_CARRIER', 
    'DELAY_DUE_WEATHER', 'DELAY_DUE_NAS', 
    'DELAY_DUE_SECURITY', 'DELAY_DUE_LATE_AIRCRAFT'
]
df_clean[delay_columns] = df_clean[delay_columns].fillna(0)

# 5. Save the cleaned dataset to a new CSV file
df_clean.to_csv('flights_final_project.csv', index=False)
```
